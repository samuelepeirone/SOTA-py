```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Computation times and speedups
### 5x5 network

```python
df = pd.read_csv('tables/5x5-1-me.csv', index_col = 0)

df
```

```python
df = pd.read_csv('tables/5x5-1-me.csv', index_col=0)
plt.figure(figsize=(10, 6))

for col in df.columns:
    plt.plot(df.index, df[col], label=col)

plt.title('Computation time for each algorithm in 5x5 grid network')
plt.xlabel('time budget')
plt.ylabel('computation time')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
df = pd.read_csv('tables/5x5-1-me-speedup.csv', index_col=0)

colors = ['#2ca02c', '#d62728', '#9467bd', '#8c564b']
plt.figure(figsize=(10, 6))

for i, col in enumerate(df.columns):
    color = colors[i % len(colors)]
    plt.plot(df.index, df[col], label=col, color=color)

plt.title('Speedups for each algorithm in 5x5 grid network')
plt.xlabel('time budget')
plt.ylabel('speedup')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```
